```python
import sqlite3, numpy as np, pandas as pd

as_of = pd.Timestamp("2026-08-01")
cust = pd.read_csv("data/customers.csv", parse_dates=["signup_date"])
txn = pd.read_sql("SELECT * FROM transactions", sqlite3.connect("data/transactions.db"))
txn["txn_timestamp"] = pd.to_datetime(txn["txn_timestamp"])

t = txn[txn.txn_timestamp < as_of].merge(cust[["customer_id", "signup_date"]], on="customer_id")
t["last_txn"] = t.groupby("customer_id").txn_timestamp.transform("max")
t["recent"] = t.txn_timestamp > t.last_txn - pd.Timedelta(days=90)   # last 3 months
t["spend"] = t.amount.where(t.is_returned == 0, 0.0)                 # returned rows excluded

g = t.groupby("customer_id")
f = pd.DataFrame({"signup_date": g.signup_date.first(), "last_txn": g.last_txn.first(),
                  "txn_all": g.size(), "ret_all": g.is_returned.sum()})

r, p = t[t.recent].groupby("customer_id"), t[~t.recent].groupby("customer_id")
f["recent_spend"], f["recent_txn"] = r.spend.sum(), r.size()
f["return_count_last_3m"] = r.is_returned.sum()
f["prev_spend"], f["prev_txn"] = p.spend.sum(), p.size()
num = ["recent_spend", "recent_txn", "return_count_last_3m", "prev_spend", "prev_txn"]
f[num] = f[num].fillna(0)

# Previous period: signup_date up to the start of the last 3 months
prev_months = ((f.last_txn - pd.Timedelta(days=90)) - f.signup_date).dt.days / 30
with np.errstate(divide="ignore", invalid="ignore"):
    f["spend_ratio"] = np.where((prev_months > 0) & (f.prev_spend > 0),
                                (f.recent_spend / 3) / (f.prev_spend / prev_months), 1.0)
    f["txn_ratio"] = np.where((prev_months > 0) & (f.prev_txn > 0),
                              (f.recent_txn / 3) / (f.prev_txn / prev_months), 1.0)

f["life_span_days"] = (f.last_txn - f.signup_date).dt.days
f["days_since_last_txn"] = (as_of - f.last_txn).dt.days
f["return_pct"] = f.ret_all / f.txn_all * 100

for d in (90, 180):
    w = t[t.txn_timestamp >= as_of - pd.Timedelta(days=d)].groupby("customer_id").size()
    f[f"txn_count_{d}d"] = w.reindex(f.index).fillna(0)

t["month"] = t.txn_timestamp.dt.to_period("M")
for m in (12, 6):
    months = pd.period_range(end=as_of.to_period("M") - 1, periods=m, freq="M")
    a = t[t.month.isin(months)].groupby("customer_id").month.nunique()
    f[f"active_months_{m}m"] = a.reindex(f.index).fillna(0)

cols = ["spend_ratio", "txn_ratio", "return_pct", "return_count_last_3m", "life_span_days",
        "days_since_last_txn", "txn_count_90d", "txn_count_180d",
        "active_months_12m", "active_months_6m"]
pd.set_option("display.width", 200)
print(f[cols].describe(percentiles=[.5, .95, .99]).T[["min", "50%", "95%", "99%", "max"]])

print("\nspend_ratio > 50 :", (f.spend_ratio > 50).sum())
print("txn_ratio > 50   :", (f.txn_ratio > 50).sum())
print("negative life_span_days (txn before signup):", (f.life_span_days < 0).sum())
print("customers with no previous period (ratio = 1.0):", (prev_months <= 0).sum())
```

```python
import sqlite3, pandas as pd

cust = pd.read_csv("data/customers.csv", parse_dates=["signup_date"])
txn = pd.read_sql("SELECT * FROM transactions", sqlite3.connect("data/transactions.db"))
agg = pd.read_parquet("data/transactions_monthly_agg.parquet")

print("=== NULLS ===")
print("customers:\n", cust.isna().sum().to_string())
print("transactions:\n", txn.isna().sum().to_string())
print("monthly_agg:\n", agg.isna().sum().to_string())

txn["txn_timestamp"] = pd.to_datetime(txn["txn_timestamp"])
agg["txn_month"] = pd.to_datetime(agg["txn_month"])

print("\n=== TRANSACTIONS ===")
print("amount skew:", round(txn.amount.skew(), 2))
print("amount mean / median:", round(txn.amount.mean(), 2), "/", round(txn.amount.median(), 2))
x = txn.merge(cust[["customer_id", "signup_date"]], on="customer_id")
print("txns before signup:", (x.txn_timestamp < x.signup_date).sum())
print("duplicate rows:", txn.duplicated().sum())

print("\n=== PARQUET ===")
print("txn_count min / max:", agg.txn_count.min(), "/", agg.txn_count.max())
print("duplicate customer-months:", agg.duplicated(["customer_id", "txn_month"]).sum())

print("\n=== PARQUET vs REBUILT ===")
txn["txn_month"] = txn.txn_timestamp.dt.to_period("M").dt.to_timestamp()
mine = (txn.groupby(["customer_id", "txn_month"], as_index=False)
           .agg(txn_count=("amount", "size"), total_amount=("amount", "sum"),
                return_count=("is_returned", "sum")))
m = mine.merge(agg, on=["customer_id", "txn_month"], how="outer",
               suffixes=("_row", "_agg"), indicator=True)
print("customer-months rebuilt from transactions.db:", len(mine))
print("customer-months in parquet:", len(agg))
print("in transactions.db only:", (m._merge == "left_only").sum())
print("in parquet only:", (m._merge == "right_only").sum())
b = m[m._merge == "both"]
cnt = b.txn_count_row != b.txn_count_agg
amt = (b.total_amount_row - b.total_amount_agg).abs() > 0.01
ret = b.return_count_row != b.return_count_agg
print("in both, count differs:", cnt.sum(), "| amount differs:", amt.sum(),
      "| return differs:", ret.sum(), "| any differs:", (cnt | amt | ret).sum())
print("transactions in db:", len(txn), "| transactions counted in parquet:", agg.txn_count.sum())
print("missing customer-months by month:")
print(m[m._merge == "left_only"].groupby("txn_month").size().to_string())

print("\n=== CUSTOMERS ACROSS SOURCES ===")
c, t, a = set(cust.customer_id), set(txn.customer_id), set(agg.customer_id)
print("csv only (zero-activity):", len(c - t))
print("in db not parquet:", len(t - a))
print("in parquet not db:", len(a - t))
```

```python
import sqlite3, numpy as np, pandas as pd

as_of = pd.Timestamp("2026-05-03")
cust = pd.read_csv("data/customers.csv", parse_dates=["signup_date"])
txn = pd.read_sql("SELECT * FROM transactions", sqlite3.connect("data/transactions.db"))
txn["txn_timestamp"] = pd.to_datetime(txn["txn_timestamp"])

elig = cust[cust.signup_date < as_of].copy()
t = txn[(txn.txn_timestamp < as_of) & txn.customer_id.isin(elig.customer_id)]
t = t.merge(elig[["customer_id", "signup_date"]], on="customer_id")

print("=== CUSTOMERS ===")
print("eligible (signup < as_of):", len(elig), "| churn rate:", round(elig.churned.mean(), 3))
print("zero-activity before as_of:", len(set(elig.customer_id) - set(t.customer_id)))
print("tenure < 365 days:", round(((as_of - elig.signup_date).dt.days < 365).mean(), 3))

t["last_txn"] = t.groupby("customer_id").txn_timestamp.transform("max")
t["recent"] = t.txn_timestamp > t.last_txn - pd.Timedelta(days=90)
t["spend"] = t.amount.where(t.is_returned == 0, 0.0)
g = t.groupby("customer_id")
f = pd.DataFrame({"signup_date": g.signup_date.first(), "last_txn": g.last_txn.first(),
                  "txn_all": g.size(), "ret_all": g.is_returned.sum()})
r, p = t[t.recent].groupby("customer_id"), t[~t.recent].groupby("customer_id")
f["recent_spend"], f["recent_txn"] = r.spend.sum(), r.size()
f["prev_spend"], f["prev_txn"] = p.spend.sum(), p.size()
f = f.fillna(0)
prev_months = ((f.last_txn - pd.Timedelta(days=90)) - f.signup_date).dt.days / 30
with np.errstate(divide="ignore", invalid="ignore"):
    f["spend_ratio"] = np.where((prev_months > 0) & (f.prev_spend > 0),
                                (f.recent_spend / 3) / (f.prev_spend / prev_months), 1.0)
    f["txn_ratio"] = np.where((prev_months > 0) & (f.prev_txn > 0),
                              (f.recent_txn / 3) / (f.prev_txn / prev_months), 1.0)
f["life_span_days"] = (f.last_txn - f.signup_date).dt.days
f["days_since_last_txn"] = (as_of - f.last_txn).dt.days
f["return_pct"] = f.ret_all / f.txn_all * 100
for d in (90, 30):
    f[f"txn_count_{d}d"] = t[t.txn_timestamp >= as_of - pd.Timedelta(days=d)] \
        .groupby("customer_id").size().reindex(f.index).fillna(0)
for d in (90, 180):
    w = t[(t.is_returned == 1) & (t.txn_timestamp > t.last_txn - pd.Timedelta(days=d))]
    f[f"return_count_{d}d"] = w.groupby("customer_id").size().reindex(f.index).fillna(0)

cols = ["spend_ratio", "txn_ratio", "return_pct", "return_count_90d", "return_count_180d",
        "life_span_days", "days_since_last_txn", "txn_count_90d", "txn_count_30d"]
pd.set_option("display.width", 200)
print("\n=== RANGES ===")
print(f[cols].describe(percentiles=[.5, .95, .99]).T[["min", "50%", "95%", "99%", "max"]])
print("no previous period (ratio = 1.0):", (prev_months <= 0).sum())

print("\n=== LEAKAGE CHECK ===")
e = elig.set_index("customer_id").join(f[["days_since_last_txn", "txn_count_90d", "ret_all"]])
rule = e.days_since_last_txn.isna() | (e.days_since_last_txn >= 90)
print("label matches 'no purchase in 90 days before as_of':", (rule == e.churned.astype(bool)).sum(), "of", len(e))
print("churn rate, txn_count_90d == 0:", round(e[e.txn_count_90d == 0].churned.mean(), 3),
      "| > 0:", round(e[e.txn_count_90d > 0].churned.mean(), 3))
print("churn rate, any returns:", round(e[e.ret_all > 0].churned.mean(), 3),
      "| no returns:", round(e[e.ret_all == 0].churned.mean(), 3))
```
